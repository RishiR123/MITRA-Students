```python
import pandas as pd
import torch
import torch.nn as nn

data = pd.read_csv("/content/Student_Performance.csv")
print(data)
```

```python
input = torch.tensor(data['Sleep Hours'].values,dtype=torch.float32).reshape(-1,1)
output =torch.tensor(data['Hours Studied'].values,dtype = torch.float32).reshape(-1,1)

model = nn.Linear(1,1)
error = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(),lr=0.001)
```

```python
for i in range(1000):
  pred = model(input)
  loss = error(pred,output)

  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
```

```python
print(model(torch.tensor([[3.0]])))
```
