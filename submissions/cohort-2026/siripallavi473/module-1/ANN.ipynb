```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/preprocessing.csv")
```

```python
df.info()
```

```python
df["label"].value_counts()
```

```python
counts=df["from"].value_counts()
print("Sender:",counts.idxmax())
print("Number of emails:",counts.max())
```

```python
avg_emails=df["thread"].value_counts()
avg=avg_emails.mean()
print(avg_emails)
print(avg)
```

```python
df["year"].unique()
```

```python
count=df["year"].value_counts()
bar_chart=plt.bar(count.index,count.values,color="pink",edgecolor="yellow")
plt.bar_label(bar_chart,fmt="%d",padding=3,label_type="edge")
plt.xticks([2024,2025,2026])
```

```python
day_count=df["day"].value_counts()
bar_chart=plt.bar(day_count.index,day_count.values,color="purple",edgecolor="yellow")
plt.bar_label(bar_chart,fmt="%d",padding=3,label_type="edge")
plt.xticks(rotation=45)
```

```python
day_count=df["day"].value_counts()
plt.pie(day_count.values,labels=day_count.index,autopct="%1.1f%%")
```

```python
mcounts=df["month"].value_counts().sort_index()
plt.plot(mcounts.index,mcounts.values,marker="o")
plt.xticks(np.arange(1,13))
```

```python
plt.hist(df["hour"],bins=24,color="green",edgecolor="red")
plt.xticks(np.arange(1,25))
```
