```python
import pandas as pd
import torch
import torch.nn as nn
```

```python
data=pd.read_csv("/content/Student_Performance (1).csv")
```

```python
data.head()
```

```python
input = torch.tensor(data['Previous Scores'].values, dtype = torch.float32).reshape(-1,1)
output = torch.tensor(data['Performance Index'].values, dtype = torch.float32).reshape(-1,1)
```

```python
model = nn.Sequential(
    nn.Linear(1, 8),
    nn.ReLU(),
    nn.Linear(8, 1)
)
```

```python
error = nn.MSELoss()
```

```python
optim = torch.optim.SGD(model.parameters(), lr = 0.001)
```

```python
for i in range(1000):
  pred = model(input)
  loss = error(pred , output)

  optim.zero_grad()
  loss.backward()
  optim.step()

  if i % 100 == 0:
    print(f"epoch {i}, loss = {loss.item():.4f}")
```
