```python
import pandas as pd
import torch
import torch.nn as nn
```

```python
# Colab: upload Student_Performance.csv, or it is next to this notebook in the repo.
import os
path = "Student_Performance.csv" if os.path.exists("Student_Performance.csv") \
       else "/content/Student_Performance.csv"

data = pd.read_csv(path)
```

```python
data
```

```python
input = torch.tensor(data['Sleep Hours'].values , dtype = torch.float32).reshape(-1,1)
output = torch.tensor(data['Hours Studied'].values, dtype = torch.float32).reshape(-1,1)
```

```python
model = nn.Linear(1,1)
```

```python
error = nn.MSELoss()
```

```python
optim = torch.optim.SGD(model.parameters(), lr = 0.001)
```

```python
for i in range(1000):
  pred = model(input)
  loss = error(pred , output)

  optim.zero_grad()
  loss.backward()
  optim.step()

```
